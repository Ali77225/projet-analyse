**Objectif de ce notebook :** charger les données avec les bons types et faire l'état des lieux avant tout nettoyage.

```python
import pandas as pd
```

```python
vente = pd.read_csv("../data/raw/online_retail.csv", parse_dates=["InvoiceDate"], dtype={"InvoiceNo": "str", "StockCode": "str", "CustomerID": "str"}, encoding="latin-1")
vente.head(15)
```

```python
vente.dtypes
```

- InvoiceNo et StockCode : Restent en str. Ce sont des identifiants, et ils contiennent des lettres
- Description : Reste str. Contient le nom des produits
- Quantity : Reste int64. Cette colonne sera utilisée afin de faire des calculs comme des sommes et des moyennes
- InvoiceDate : str ==> date. Afin d'effectuer une analyse par période
- UnitPrice : Reste float64. Un prix se calcule avec des décimales
- CustomerID : float64 ==> str. Il s'agit d'un identifiant client, il ne sert pas pour les calculs
- Country : Reste str. Va servir pour grouper
